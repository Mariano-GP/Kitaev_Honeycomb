```python
import numpy as np

N = 22
adj_x = np.zeros((N, N), dtype=int)
adj_y = np.zeros((N, N), dtype=int)
adj_z = np.zeros((N, N), dtype=int)

z_bonds = [(1,18),(3,4),(5,6),(7,8),(10,22),(12,21),(14,15),(16,17),(19,20)]
x_bonds = [(1,2),(4,19),(6,22),(8,9),(10,11),(12,13),(15,16),(17,18),(20,21)]
y_bonds = [(2,3),(4,5),(6,7),(9,10),(11,12),(13,14),(16,21),(18,19),(20,22)]

for (i, j) in z_bonds:
    adj_z[i-1, j-1] = adj_z[j-1, i-1] = 1
for (i, j) in x_bonds:
    adj_x[i-1, j-1] = adj_x[j-1, i-1] = 1
for (i, j) in y_bonds:
    adj_y[i-1, j-1] = adj_y[j-1, i-1] = 1

np.save("Jz.npy", adj_z)
np.save("Jx.npy", adj_x)
np.save("Jy.npy", adj_y)
```
